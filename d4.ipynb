```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
import joblib
```

```python
data={
    'marks':[10,20,30,40,50,60,70,80,],
    'result':[0,0,0,0,1,1,1,1,]
    
}
```

```python
df=pd.DataFrame(data)
x=df[['marks']]
y=df[['result']]
model = LogisticRegression()
model.fit(x,y)
joblib.dump(model,'my_marks_model.pkl')
```
