```python
from openai import OpenAI
import gradio as gr
import os
import time

# Setup Gemini API
api_key = os.getenv("GEMINI_API_KEY", "AIzaSyBVzYPGp-4HF5CaYq0yvR7jWN3g_i_Hhwg")

gemini_model = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

# AI Assistant function with comprehensive error handling
def ai_assistant(user_prompt, assistant_type="general"):
    if not user_prompt.strip():
        return "Please enter a question or prompt to get started!"
    
    try:
        # Different system prompts based on assistant type
        system_prompts = {
            "general": "You are a helpful AI assistant. Provide clear, accurate, and useful responses to user questions.",
            "creative": "You are a creative AI assistant. Help users with creative writing, brainstorming, and imaginative tasks. Be inspiring and original.",
            "technical": "You are a technical AI assistant. Provide detailed, accurate technical information and solutions. Be precise and thorough.",
            "study": "You are an educational AI tutor. Help users learn and understand concepts clearly. Break down complex topics into simple explanations.",
            "business": "You are a business AI consultant. Provide professional advice on business strategies, marketing, and entrepreneurship."
        }
        
        system_content = system_prompts.get(assistant_type, system_prompts["general"])
        
        messages = [
            {"role": "system", "content": system_content},
            {"role": "user", "content": user_prompt}
        ]
        
        # Make API call with timeout handling
        response = gemini_model.chat.completions.create(
            model="gemini-1.5-flash",
            messages=messages,
            max_tokens=1000,
            temperature=0.7
        )
        
        if response and response.choices:
            return response.choices[0].message.content.strip()
        else:
            return "Sorry, I couldn't generate a response. Please try again."
            
    except Exception as e:
        error_msg = str(e).lower()
        
        # Handle specific error types
        if "api_key" in error_msg or "authentication" in error_msg:
            return "❌ API Key Error: Please check your Gemini API key configuration."
        elif "rate_limit" in error_msg or "quota" in error_msg:
            return "⏳ Rate Limit: Too many requests. Please wait a moment and try again."
        elif "timeout" in error_msg or "connection" in error_msg:
            return "🌐 Connection Error: Please check your internet connection and try again."
        elif "model" in error_msg:
            return "🤖 Model Error: There's an issue with the AI model. Please try again later."
        else:
            return f"⚠ Unexpected Error: {str(e)[:100]}... Please try again or contact support."

# Create the Gradio interface
def create_interface():
    with gr.Blocks(theme=gr.themes.Soft(), title="AI Assistant Hub") as iface:
        gr.Markdown("# 🤖 AI Assistant Hub")
        gr.Markdown("Choose your AI assistant type and get intelligent responses!")
        
        with gr.Row():
            with gr.Column(scale=3):
                user_input = gr.Textbox(
                    label="Your Question/Prompt",
                    placeholder="Ask me anything... I'm here to help!",
                    lines=3
                )
                
                assistant_type = gr.Dropdown(
                    choices=["general", "creative", "technical", "study", "business"],
                    value="general",
                    label="Assistant Type",
                    info="Choose the type of assistance you need"
                )
                
                submit_btn = gr.Button("Get Response", variant="primary", size="lg")
                clear_btn = gr.Button("Clear", variant="secondary")
                
            with gr.Column(scale=4):
                output = gr.Textbox(
                    label="AI Response",
                    placeholder="Your AI response will appear here...",
                    lines=10,
                    max_lines=15
                )
        
        # Example prompts
        gr.Markdown("### 💡 Example Prompts:")
        with gr.Row():
            example_btns = [
                gr.Button("Explain quantum physics", size="sm"),
                gr.Button("Write a short story", size="sm"),
                gr.Button("Business plan ideas", size="sm"),
                gr.Button("Study tips", size="sm")
            ]
        
        # Event handlers
        submit_btn.click(
            fn=ai_assistant,
            inputs=[user_input, assistant_type],
            outputs=output
        )
        
        clear_btn.click(
            fn=lambda: ["", ""],
            outputs=[user_input, output]
        )
        
        # Example button handlers
        example_btns[0].click(lambda: "How does quantum physics work in simple terms?", outputs=user_input)
        example_btns[1].click(lambda: "Write a creative short story about time travel", outputs=user_input)
        example_btns[2].click(lambda: "Give me 5 innovative business ideas for 2024", outputs=user_input)
        example_btns[3].click(lambda: "What are the best study techniques for better learning?", outputs=user_input)
        
        # Allow Enter key to submit
        user_input.submit(
            fn=ai_assistant,
            inputs=[user_input, assistant_type],
            outputs=output
        )
    
    return iface

# Launch the application
if _name_ == "_main_":
    interface = create_interface()
    interface.launch(
        share=True,
        server_name="0.0.0.0",
        server_port=7860,
        show_error=True
    )from openai import OpenAI
import gradio as gr
import os
import time

# Setup Gemini API
api_key = os.getenv("GEMINI_API_KEY", "AIzaSyCFGoqRZGdoPC0JxSjmsTtfn8ngu2gO8ZU")

gemini_model = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta2/openai/"
)

# AI Assistant function with comprehensive error handling
def ai_assistant(user_prompt, assistant_type="general"):
    if not user_prompt.strip():
        return "Please enter a question or prompt to get started!"
    
    try:
        # Different system prompts based on assistant type
        system_prompts = {
            "general": "You are a helpful AI assistant. Provide clear, accurate, and useful responses to user questions.",
            "creative": "You are a creative AI assistant. Help users with creative writing, brainstorming, and imaginative tasks. Be inspiring and original.",
            "technical": "You are a technical AI assistant. Provide detailed, accurate technical information and solutions. Be precise and thorough.",
            "study": "You are an educational AI tutor. Help users learn and understand concepts clearly. Break down complex topics into simple explanations.",
            "business": "You are a business AI consultant. Provide professional advice on business strategies, marketing, and entrepreneurship."
        }
        
        system_content = system_prompts.get(assistant_type, system_prompts["general"])
        
        messages = [
            {"role": "system", "content": system_content},
            {"role": "user", "content": user_prompt}
        ]
        
        # Make API call with timeout handling
        response = gemini_model.chat.completions.create(
            model="gemini-1.5-flash",
            messages=messages,
            max_tokens=1000,
            temperature=0.7
        )
        
        if response and response.choices:
            return response.choices[0].message.content.strip()
        else:
            return "Sorry, I couldn't generate a response. Please try again."
            
    except Exception as e:
        error_msg = str(e).lower()
        
        # Handle specific error types
        if "api_key" in error_msg or "authentication" in error_msg:
            return "❌ API Key Error: Please check your Gemini API key configuration."
        elif "rate_limit" in error_msg or "quota" in error_msg:
            return "⏳ Rate Limit: Too many requests. Please wait a moment and try again."
        elif "timeout" in error_msg or "connection" in error_msg:
            return "🌐 Connection Error: Please check your internet connection and try again."
        elif "model" in error_msg:
            return "🤖 Model Error: There's an issue with the AI model. Please try again later."
        else:
            return f"⚠ Unexpected Error: {str(e)[:100]}... Please try again or contact support."

# Create the Gradio interface
def create_interface():
    with gr.Blocks(theme=gr.themes.Soft(), title="AI Assistant Hub") as iface:
        gr.Markdown("# 🤖 AI Assistant Hub")
        gr.Markdown("Choose your AI assistant type and get intelligent responses!")
        
        with gr.Row():
            with gr.Column(scale=3):
                user_input = gr.Textbox(
                    label="Your Question/Prompt",
                    placeholder="Ask me anything... I'm here to help!",
                    lines=3
                )
                
                assistant_type = gr.Dropdown(
                    choices=["general", "creative", "technical", "study", "business"],
                    value="general",
                    label="Assistant Type",
                    info="Choose the type of assistance you need"
                )
                
                submit_btn = gr.Button("Get Response", variant="primary", size="lg")
                clear_btn = gr.Button("Clear", variant="secondary")
                
            with gr.Column(scale=4):
                output = gr.Textbox(
                    label="AI Response",
                    placeholder="Your AI response will appear here...",
                    lines=10,
                    max_lines=15
                )
        
        # Example prompts
        gr.Markdown("### 💡 Example Prompts:")
        with gr.Row():
            example_btns = [
                gr.Button("Explain quantum physics", size="sm"),
                gr.Button("Write a short story", size="sm"),
                gr.Button("Business plan ideas", size="sm"),
                gr.Button("Study tips", size="sm")
            ]
        
        # Event handlers
        submit_btn.click(
            fn=ai_assistant,
            inputs=[user_input, assistant_type],
            outputs=output
        )
        
        clear_btn.click(
            fn=lambda: ["", ""],
            outputs=[user_input, output]
        )
        
        # Example button handlers
        example_btns[0].click(lambda: "How does quantum physics work in simple terms?", outputs=user_input)
        example_btns[1].click(lambda: "Write a creative short story about time travel", outputs=user_input)
        example_btns[2].click(lambda: "Give me 5 innovative business ideas for 2024", outputs=user_input)
        example_btns[3].click(lambda: "What are the best study techniques for better learning?", outputs=user_input)
        
        # Allow Enter key to submit
        user_input.submit(
            fn=ai_assistant,
            inputs=[user_input, assistant_type],
            outputs=output
        )
    
    return iface

# Launch the application
if _name_ == "_main_":
    interface = create_interface()
    interface.launch(
        share=True,
        server_name="0.0.0.0",
        server_port=7860,
        show_error=True
    )
```
